```python
import pandas as pd
import numpy as np
from surprise import Reader,Dataset,SVD
import pickle
import os
```

```python
modify_path = "../CineIq_Data/modify/"
ratings_path = "../CineIq_Data/ml-25m/ratings.csv"
movies_path  = "../CineIq_Data/modify/cineiq_metadata.csv"
ratings = pd.read_csv(ratings_path).drop(columns=["timestamp"]).sample(n=1000000, random_state=42)
movies = pd.read_csv(movies_path)[["movieId","title"]]
```

```python
ratings.head()
```

```python
movies.head()
```

```python
reader = Reader(rating_scale=(0.5,5.0))
data = Dataset.load_from_df(ratings,reader)
trainset = data.build_full_trainset()
svd_model = SVD(n_factors=100, random_state=42)
svd_model.fit(trainset)
```

```python
os.makedirs(modify_path, exist_ok=True)
with open(modify_path + "svd_model.pkl", "wb") as f:
    pickle.dump(svd_model,f)
```
