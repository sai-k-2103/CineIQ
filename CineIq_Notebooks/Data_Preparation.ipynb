```python
import pandas as pd
import ast
```

```python
ml_path = "../CineIq_Data/ml-25m/"
tmdb_path = "../CineIq_Data/tmdb-45k/"
```

```python
ml_movies = pd.read_csv(ml_path + "movies.csv",)
ml_links = pd.read_csv(ml_path + "links.csv")

ml_merged = pd.merge(ml_movies,ml_links,on="movieId")
ml_merged = ml_merged.dropna(subset=["tmdbId"])
ml_merged["tmdbId"] = ml_merged["tmdbId"].astype(int)
```

```python
tmdb_meta = pd.read_csv(tmdb_path + "movies_metadata.csv",low_memory=False)
tmdb_meta["id"] = pd.to_numeric(tmdb_meta["id"],errors="coerce")
tmdb_meta = tmdb_meta.dropna(subset=["id"])
tmdb_meta["id"] = tmdb_meta["id"].astype(int)
```

```python
tmdb_keywords = pd.read_csv(tmdb_path + "keywords.csv")
tmdb_keywords["id"] = pd.to_numeric(tmdb_keywords["id"],errors="coerce")
tmdb_keywords = tmdb_keywords.dropna(subset=["id"])
tmdb_keywords["id"] = tmdb_keywords["id"].astype(int)
```

```python
base_df = pd.merge(ml_merged, tmdb_meta, left_on="tmdbId",right_on="id",how="inner")
final_df = pd.merge(base_df, tmdb_keywords, on="id",how="inner")
```

```python
final_df.columns
```

```python
required_columns = ["movieId","title_x","genres_y","overview","keywords"]
cineiq_df = final_df[required_columns].rename(columns={
    "title_x": "title",
    "genres_y": "tmdb_genres",
    "keywords": "tmdb_keywords",
    "overview": "tmdb_overview"
})
```

```python
cineiq_df.head()
```

```python
def extract_names(x):
    if(pd.isna(x)):
        return ""
    try:
        data = ast.literal_eval(x)
        if isinstance(data,list):
            return " ".join([i["name"] for i in data])
    except:
        pass
    return ""
```

```python
cineiq_df["genres"] = cineiq_df["tmdb_genres"].apply(extract_names)
cineiq_df["keywords"] = cineiq_df["tmdb_keywords"].apply(extract_names)
cineiq_df["overview"] = cineiq_df["tmdb_overview"].fillna('')
```

```python
cineiq_df = cineiq_df[["movieId","title","overview","genres","keywords"]]
```

```python
output_path = "../CineIq_Data/modify/cineiq_metadata.csv"
cineiq_df.to_csv(output_path,index=False)
```
